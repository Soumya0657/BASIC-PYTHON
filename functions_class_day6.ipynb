```python
#make a function and convert list to dictionary

def list_to_dict(lst):

    if len(lst) % 2 != 0:
        raise ValueError("List must contain an even number of elements to form key-value pairs.")

    return dict(zip(lst[::2], lst[1::2]))


my_list = [1,'a', 2, 'b',  3, 'c']
result = list_to_dict(my_list)
print(result)
```
