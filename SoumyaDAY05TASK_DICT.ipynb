```python
#invert a inverted dictionary
my_map = { 'a': 1, 'b':2 }
dict(map(reversed, my_map.items()))
```

```python
#filter Dictionary Values
dict = {'a':5, 'b':2}
x = 4;
print ([dict[i] for i in dict if dict[i] >= x])


```

```python
#Combine 2 lists to create a dictionary
dict(zip([1,2,3,4], ['a', 'b', 'c', 'd']))
```

```python
# Count the characters of a string
s = "TECHNICAL OFFICER!"


count = len(s)

print(count)
```

```python
# Square the dictionary values
my_dict = {'a': 2, 'b': 3, 'c': 4}


squared_dict = {key: value ** 2 for key, value in my_dict.items()}

print(squared_dict)

```
