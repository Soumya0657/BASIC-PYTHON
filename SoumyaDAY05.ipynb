```python
#multiplication table of 1 2 3
for i in range(1,4):
  for j in range(1,11):
    print(f"{i}*{j}={i*j}")
    print()
```

```python
#String filter len<5 remove
stss=["Apple" , "Banana"]

filtered=[s for s in stss if len(s)>5]

print(filtered)
```
